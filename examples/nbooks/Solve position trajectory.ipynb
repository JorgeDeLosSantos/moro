```python
import numpy as np
import moro as mr

from moro.abc import q1, q2, l1, l2
from moro.inverse_kinematics import solve_position_trajectory
from moro.visualization import RobotVisualizer, VisualizationStyle

# Robot RR simbólico
robot = mr.Robot(
    (l1, 0, 0, q1, "r"),
    (l2, 0, 0, q2, "r"),
)

# Valores numéricos para parámetros geométricos
params = {l1: 200, l2: 200}

# Secuencia de objetivos cartesianos (m, 3)
targets = [[t,220,0] for t in np.linspace(-100,100)]

# Semilla inicial obligatoria para el primer punto
q0 = [0.5, 0.5]

trajectory = solve_position_trajectory(
    robot,
    targets,
    q0=q0,
    parameters=params,
    method="ccd",
    tol=1e-8,
    max_iter=200,
)

print("Converged:", trajectory.converged)
print("Failed index:", trajectory.failed_index)
print("Message:", trajectory.message)
# print("Joint solutions:", trajectory.qs)
# print("Per-point errors:", trajectory.errors)
# print("Per-point iterations:", trajectory.iterations)

if not trajectory.converged:
    raise RuntimeError("La trayectoria no convergió: " + trajectory.message)

# Uso directo de trajectory.qs para visualización/animación
# Convertimos cada q en un diccionario de sustitución simbólica
num_vals_list = []
for q in trajectory.qs:
    num_vals_list.append({
        q1: q[0],
        q2: q[1],
        l1: params[l1],
        l2: params[l2],
    })

viz = RobotVisualizer(robot)
viz.animate(num_vals_list, backend="threejs", style=VisualizationStyle(show_trajectory=True))
```
