```python
import numpy as np
import matplotlib.pyplot as plt
from moro import Robot
from moro.visualization import RobotVisualizer, VisualizationStyle
from moro.abc import q1,q2,q3,l1,l2,l3,d1
from sympy import pi
from IPython.display import HTML
```

## RR Robot visualization

```python
rr = Robot((l1, 0, 0, q1, "r"), (l2, 0, 0, q2, "r"))
viz = RobotVisualizer(rr)
```

Using Three.js backend:

```python
viz.plot({q1:np.pi/6, q2:np.pi/6, l1:100, l2:100}, backend="threejs")
```

Using Matplotlib backend:

```python
viz.plot({q1:np.pi/6, q2:np.pi/6, l1:100, l2:100}, backend="matplotlib")
```

## RR Robot Animation

```python
positions = [{q1:np.sin(t), q2:t, l1:100, l2:100} for t in np.linspace(0,2*np.pi,50)]
viz.animate(positions, backend="threejs")
```
