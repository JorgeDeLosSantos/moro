```python
import moro as mr
from moro import l1,l2,q1,q2
```

Creating the manipulator using the `Robot` class.

```python
RR = mr.Robot(
    (l1,0,0,q1,"r"),
    (l2,0,0,q2,"r")
)
```

Getting the Jacobian matrix using the `J` attribute:

```python
RR.J
```
