```python
import moro as mr
from moro.abc import q1,q2,q3,d1,l2,l3
import sympy as sp
mr.init_vprinting()
```

```python
RRR = mr.Robot(
    (0,sp.pi/2,d1,q1),
    (l2,0,0,q2),
    (l3,0,0,q3)
)
```
